```python
import numpy as np
import plotly.express as px
import pandas as pd

f_in = 10e3
```

```python
timebase = np.linspace(0, 1/f_in, 10000)
signal = np.sin(2*np.pi*f_in*timebase)

reference = 1
integrator_1 = 0
feedback =0
integrator_2 = 0
bitstream = []

for val in signal:
    integrator_1 += val - feedback
    integrator_2 += integrator_1
    comparator_in = integrator_2 + integrator_1 + val
    if comparator_in > reference:
        feedback = reference
        bitstream.append(1)
    else:
        feedback = -reference
        bitstream.append(-1)

df = pd.DataFrame({"in": signal, "out": bitstream}, index=timebase)
```

```python
px.line(df)
```
