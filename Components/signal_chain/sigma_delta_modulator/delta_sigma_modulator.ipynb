```python
import numpy as np
import plotly.express as px
import pandas as pd

f_in = 10e3
```

```python
sim_len = 10000
timebase = np.linspace(0, 1/f_in, sim_len)
signal = [18000 for x in range(0, sim_len)] #np.sin(2*np.pi*f_in*timebase)

reference = 32767
feedback =0
bitstream = [0]*10000
integrators = [[0]*10000, [0]*10000]
comparator_in = [0]*10000

for i in range(1, len(signal)):
    integrators[0][i] =  integrators[0][i-1] +  signal[i] - bitstream[i-1]
    integrators[1][i] = integrators[1][i-1] + integrators[0][i]
    comparator_in[i] = integrators[1][i] + integrators[0][i] + signal[i]
    if comparator_in[i] > reference:
        bitstream[i] = reference
    else:
        bitstream[i] = -reference


df = pd.DataFrame({"in": signal, "out": bitstream, "integrator_1": integrators[0], "integrator_2": integrators[1]}, index=timebase)
```

```python
px.line(df, line_shape="vh")
```
